```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/MAPCO2_SAMI.csv')
print(df.columns)
```

```python
df['date_str'] = df['date'].astype(str)
df['time_str'] = df['time'].astype(str)

# Create a datetime column by combining year, month, day, and time (assuming 'HHMM' format)
df['datetime'] = pd.to_datetime(df['date_str'] + df['time_str'])

# Set desired format with timezone offset (-05:00 in this example)
df['created_at'] = df['datetime'].dt.strftime('%Y-%m-%dT%H:%M:%S+00:00')

# Print the result (excluding unnecessary columns for clarity)
print(df[['created_at']].to_string(index=False))
```

```python
print(df.columns)
```

```python
# Desired columns (including datetime_str assuming it's created as mentioned previously)
chosen_columns = ['created_at', 'pCO2_5m', 'pH_1m', 'Temp_5m']  # Adjust as needed

# Create a new DataFrame with chosen columns
new_df = df[chosen_columns]

# Reorder columns to make 'datetime_str' the first column
data = new_df[['created_at'] + chosen_columns[1:]]  # Select datetime_str first, then the rest

# Print the new DataFrame
print(data)
```

# Data Cleaning

## Removing Missing Values

```python
# Replace "nd" with NaN
data.replace("nd", float("NaN"), inplace=True)

# Handle missing values
# For example, fill missing values with linear interpolation
#data.interpolate(method='linear', inplace=True)

# Alternatively, to delete rows with missing values:
data.dropna(inplace=True)

# Display the updated dataset
print(data)
```

    Rolling Averages: By smoothing out short-term fluctuations, rolling averages provide a clearer picture of the underlying trends in the data Anomalies that cause abrupt changes or deviate from these smoothed trends are more easily detectable.

    Differences Between Consecutive Measurements: Differences between consecutive measurements highlight the rate of change in the sensor data. Anomalies that cause sudden spikes or drops in measurements are often reflected in these differences.

```python
# Convert pH_1m column to numeric data type
data['pH_1m'] = pd.to_numeric(data['pH_1m'], errors='coerce')

# Calculate rolling averages
window_size = 3
data['pCO2_5m_rolling_mean'] = data['pCO2_5m'].rolling(window=window_size).mean()
data['pH_1m_rolling_mean'] = data['pH_1m'].rolling(window=window_size).mean()
data['Temp_5m_rolling_mean'] = data['Temp_5m'].rolling(window=window_size).mean()

# Calculate differences between consecutive measurements
data['pCO2_5m_diff'] = data['pCO2_5m'].diff()
data['pH_1m_diff'] = data['pH_1m'].diff()
data['Temp_5m_diff'] = data['Temp_5m'].diff()

# Display the updated dataset with additional features
print(data.head())
```
